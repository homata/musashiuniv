```python
# カフェの1週間の売上データ
daily_sales = np.array()

# 「10000円より多くて（>）、かつ（&）、20000円未満（<）」の日だけ抽出したい！
# NumPyでは、and の代わりに 「 & 」を使い、条件ごとに ( ) で囲むルールがあります。
normal_days = daily_sales[(daily_sales > 10000) & (daily_sales < 20000)]

print("条件に合う売上データ:", normal_days)
```
