```python
# 1日の注文履歴（辞書のリスト）
daily_orders = [
    {"item": "コーヒー", "price": 450},
    {"item": "ケーキ", "price": 500},
    {"item": "コーヒー", "price": 450}
]

# 合計金額を入れる箱（最初は0円）
total_sales = 0

# for文で1つずつ注文を取り出して、値段を足していく
for order in daily_orders:
    total_sales = total_sales + order["price"]

print("本日の合計売上は", total_sales, "円です！")
```
