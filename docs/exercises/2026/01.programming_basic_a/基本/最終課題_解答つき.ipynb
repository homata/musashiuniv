## 課題7
1から20までの整数を順番に出力します。ただし、3の倍数の場合は数の代わりに”Fizz”、5の倍数の場合は”Buzz”、3と5の公倍数の場合は”FizzBuzz”と出力してください。

**期待する出力：**   
True  
True  
True  
True  
True  
True  


````python
def fizzbuzz(i):
    if i % 3 == 0 and i % 5 == 0:
        return "FizzBuzz"
    elif i % 3 == 0:
        return "Fizz"
    elif i % 5 == 0:
        return "Buzz"
    else:
        return i

print(fizzbuzz(3)=="Fizz")
print(fizzbuzz(5)=="Buzz")
print(fizzbuzz(15)=="FizzBuzz")
print(fizzbuzz(225)=="FizzBuzz")
print(fizzbuzz(13)==13)
print(fizzbuzz(2)==2)
````

## 課題8
height(身長 (cm))、weight(体重 (kg))を引数として、次の式によってBMI（ボディマス指数）を計算する式を完成させて、BMIの標準値の範囲内（18.5～25未満）かをチェックしてください．

```
BMI ＝ 体重(kg) ÷ (身長(m)) x (身長(m))
```

**期待する出力：** True

````python
def bmi(height, weight):
  return weight / (height/100.0) ** 2

height = 180 # 180cm
weight = 60  # 60kg
ans = bmi(height, weight)
print(18.5 <= ans and ans < 25)
````

## 課題9

整数値を値にもつ1次元配列を引数として、5の倍数かつ2で割って1余る数を要素として持つ配列を出力する関数を実装してください。

**期待する出力：** True

````python
import numpy as np

def answer_ten(a):
    idx = (a % 5 == 0) & (a % 2 == 1)
    return a[idx]

# Numpyの1次元配列の定義
a = np.array([1, 5, 10, 3, 4, 25, 30])
array = answer_ten(a)
print(array[0] == 5 and array[1] == 25)
````

## 課題10
次のDataFrame `df` はある電化製品店のパソコンコーナーの受注情報の一部です。  
このDataFrame(df)に格納されている一番最初の顧客の名前は何ですか？

| column  |  description  |
|---------|------------------------------|
|  id     | 顧客番号 |
|  name   | 顧客名 |
|  date   | 来店日 |
|  laptop | 購入または修理に持ち込んだパソコンの種類 |
|  fix    | 修理受注であればTrue、購入であればFalse |


**期待する出力：** True

````python
import numpy as np
import pandas as pd

def make_df():
    df = pd.DataFrame({'id':[100, 100, 101, 101, 101, 102, 103, 104, 104, 105],
                      'name': ['Max', 'Max', 'Bob', 'Bob', 'Bob', 'Alice', 'Chris','Emma', 'William','Jacob'],
                       'gender':['m', 'm', 'm', 'm', 'm', 'f', np.nan, 'f', 'm', 'm'],
                       'date':['2019/3/4','2019/2/5','2018/10/3','2017/12/6','2016/9/3','2019/7/4','2017/1/1','2019/5/3','2018/2/18', '2015/12/30'],
                      'laptop':['mac','mac','windows','windows','mac', 'mac', 'windows', 'windows', 'mac', 'others'],
                      'fix': [True, False, True, False, False, False, False, True, True, True],
                      'price': [10000, 230000, 5400, 15000, 300000, 240000, 180000, 150000, 100000, 3000]})
    return df

def answer_eleven():
    df = make_df()
    ans = df.loc[0,'name']
    return ans

print(answer_eleven()=='Max')
````
