```python
# --- ④ ヒストグラム ---
# お客様50人分の年齢データ（ダミー）を生成します
np.random.seed(0)
ages = np.random.randint(15, 65, 50) # 15歳〜65歳の間でランダムに50個

# seabornを使ってヒストグラムを描画（kde=True をつけると、なめらかな山の線も描いてくれます）
sns.histplot(ages, bins=10, kde=True, color="green")
plt.title("ご来店のお客様の年齢分布")
plt.xlabel("年齢")
plt.ylabel("人数")
plt.show()

# ========================================================
```
