```python
# カフェの設計図（クラス）を作ります
class Cafe:

    # __init__ は、お店を建てる時（初期化）に最初に呼ばれる特別なメソッドです。
    def __init__(self, name):
        # 自分自身(self)の name属性 に、渡された名前を記録します
        self.name = name
        self.sales = 0

    # お店ができること（メソッド）を定義します
    def open_shop(self):
        print(self.name, "がオープンしました！いらっしゃいませ！")

# ----------------------------------------------------
# 設計図(Cafeクラス)を使って、実際の店舗（インスタンス）を作ります！
tokyo_cafe = Cafe("東京本店")
osaka_cafe = Cafe("大阪支店")

# 属性（データ）にアクセスする
print("店舗名:", tokyo_cafe.name)

# メソッド（機能）を呼び出す
tokyo_cafe.open_shop()
osaka_cafe.open_shop()
```

```python
# 例外処理の基本
def divide_sales(sales, people):
    try:
        # まずは普通に割り算を試します
        result = sales / people

    except ZeroDivisionError:
        # もし people が 0 で 0割りエラーが起きたら、止まらずにここに来ます！
        print("【エラー】人数が0人です！計算できません。")
        result = 0

    else:
        # エラーなく無事に計算できた時だけ呼ばれます
        print("無事に計算完了しました。")

    finally:
        # エラーが起きても起きなくても、絶対に最後に呼ばれます
        print("--- 処理終了 ---")

    return result

# 正常なデータ
print("1人あたり:", divide_sales(1000, 2), "円\n")

# 異常なデータ（人数が0人！ 普通ならプログラムがクラッシュします）
print("1人あたり:", divide_sales(1000, 0), "円")
```

```python
# 正しい数値リストが完成するまで、エラーを無視して処理を続けるコード
data_list = [10, "A", 20, "B", 30]
valid_numbers = []

for item in data_list:
    try:
        # 数値への変換を試みる
        num = int(item)
        valid_numbers.append(num)
        print("追加成功:", num)
    except ValueError:
        # 文字が来てエラーになったら、警告だけ出して処理は止めない！
        print("【警告】数値ではないデータをスキップしました:", item)

print("最終的な正しいデータ:", valid_numbers)
```
