# 第4回 数値演算と文字列



```python
# 簡単な算術計算と優先順位の確認
# 400円のコーヒーと300円のクッキーを買い、100円引きクーポンを使いました。

# 悪い例：読みにくい（空白がない）
total1=400+300-100

# 良い例：読みやすい（空白がある）
total2 = 400 + 300 - 100

print("合計金額:", total2)

# 括弧を使った優先順位の変更
# 3人でカフェに行き、各自が400円と300円のものを頼みました
wrong_calc = 400 + 300 * 3    # 300*3が先に計算されてしまう（400 + 900 = 1300）
correct_calc = (400 + 300) * 3 # 括弧の中が先に計算される（700 * 3 = 2100）

print("間違った計算:", wrong_calc)
print("正しい計算:", correct_calc)
```

```python
# 数学関数が入っている「math（マス）」モジュールを取り込みます
import math

# 小数点以下を「切り上げ」る関数 math.ceil()
# 例：ポイント還元で 1.2 ポイントの時は 2 ポイントに切り上げる
points = math.ceil(1.2)
print("切り上げポイント:", points)

# 小数点以下を「切り捨て」る関数 math.floor()
price = math.floor(450.9)
print("切り捨て価格:", price)
```

```python
# 練習：来店者番号が「3の倍数」の人をキャンペーン対象（当たり）とします。
# 割り算の余りが 0 なら 3の倍数です！
customer_id = 12

# customer_id を 3 で割った余りを計算
remainder = customer_id % 3
print("余りは", remainder, "です。0なら当たり！")
```

```python
shop_name = "ムサシ・カフェ"

# インデックス（1文字だけ取り出す）
# 0から数えるので、3番目は「・」ではなく「カ」になります！
print("0番目の文字:", shop_name)
print("3番目の文字:", shop_name)
print("最後の文字:", shop_name[-1])

# スライス（範囲を切り出す）
# 0番目から4番目の「手前（3番目）」まで切り出す
print("0から4番目の手前まで:", shop_name[0:4])

# 開始や終了を省略すると、「一番最初から」「一番最後まで」という意味になります
print("最初から3番目の手前まで:", shop_name[:3])
print("4番目から最後まで:", shop_name[4:])
```

```python
# 文字列の連結
greeting = "こんにちは！"
customer = "山田さん"
print(greeting + customer) # こんにちは！山田さん

# エスケープシーケンス（\n で改行）
receipt = "コーヒー：400円\nケーキ：500円\n合計：900円"
print(receipt)
```

```python
# 検索（in）
menu = "濃厚抹茶フラペチーノ"
print("「抹茶」は含まれている？:", "抹茶" in menu)

# 比較（==）
password = "open"
print("パスワードは正しい？:", password == "open")
```

```python
text = "  Cafe Musashi  "

# upper(): すべて大文字にする
print("大文字:", text.upper())

# replace(): 文字を置き換える（データクレンジングで超頻出！）
print("置換:", text.replace("Musashi", "Tokyo"))

# strip(): 前後の不要な空白を削除する
print("空白削除:[", text.strip(), "]")
```

```python
drink = "コーヒー"  # 変数 drink に 文字列 "コーヒー" を入れた

print("drink")    # ← 画面には「drink」という文字がそのまま出る！
print(drink)      # ← 画面には箱の中身の「コーヒー」が出る！
```
