```python
# for文によるリストの繰り返し
sales_list =  # コーヒー、ケーキ、紅茶の売上

# リストから1つずつ取り出して、変数 price に入れながら繰り返す！
for price in sales_list:
    print("売上データ:", price, "円")

# for文による文字列の繰り返し（文字も1文字ずつ取り出せます！）
for char in "CAFE":
    print("文字:", char)
```

```python
# range() 関数（0から4までの整理券を発行する）
for i in range(5):
    print("整理券番号:", i)

# rangeとリストの組み合わせ
num_list = list(range(1, 4))
print("作成されたリスト:", num_list)

# enumerate() で番号札とデータを両方取り出す！
menu = ["コーヒー", "紅茶", "ケーキ"]
for index, item in enumerate(menu):
    print(index, "番目のメニューは", item, "です。")
```

```python
# 辞書の繰り返し
prices = {"コーヒー": 400, "紅茶": 400}
# .items() を使うと、キーと値を両方一気に取り出せます
for key, value in prices.items():
    print(key, "の値段は", value, "円です。")

# for文の入れ子（九九の計算のようなイメージ）
for i in range(1, 3):
    for j in range(1, 3):
        print("i=", i, ", j=", j)
```

```python
# while文 と break, continue の働き
stock = 3 # 在庫

while stock > 0:
    print("いらっしゃいませ！在庫は残り", stock, "個です。")
    stock = stock - 1

    if stock == 1:
        print("残り1個なので、特別なお客様用にとっておきます！（スキップ）")
        continue # ここで下の処理を飛ばして、次のループの先頭に戻る

    if stock == 0:
        print("売り切れました！本日は閉店します。（強制終了）")
        break # ループを完全に抜け出す

# ループの else（無事に完走したかどうかの確認）
for i in range(3):
    print("処理中...", i)
    # ここに break があると、下の else は実行されません
else:
    print("ループがエラーや中断なく、無事に最後まで完了しました！")
```

```python
# イテラブルとイテレータの違いを体験してみましょう！
my_list = ["コーヒー", "紅茶", "ケーキ"] # これはイテラブル（箱）です

# iter() で、この箱専用の「取り出し係（イテレータ）」を作ります
my_iterator = iter(my_list)

# next() を呼ぶたびに、現在地を記憶しながら1つずつ出してくれます
print("1回目:", next(my_iterator))
print("2回目:", next(my_iterator))
print("3回目:", next(my_iterator))
# 4回目を呼ぶと、空っぽなので StopIteration というエラーが出ます！

```

```python
# 通常のfor文で税込み価格のリストを作る場合
prices =
tax_in_prices = []
for p in prices:
    tax_in_prices.append(p * 1.1)
print("通常のfor文:", tax_in_prices)

# 🌟 リスト内包表記を使うと、これがたった1行で書けます！
# 読み方：「pricesの中の p を一つずつ取り出し、p * 1.1 をした結果をリストにまとめる」
tax_in_prices_fast = [p * 1.1 for p in prices]
print("リスト内包表記:", tax_in_prices_fast)

# 条件付き内包表記（200円以上のものだけ税込みにする）
expensive_only = [p * 1.1 for p in prices if p >= 200]
print("条件付き（200円以上）:", expensive_only)

# 辞書内包表記の例（メニュー名をキー、値段を値にする）
menu_names = ["コーヒー", "紅茶", "ケーキ"]
# zip は2つのリストを同時に取り出す便利ツールです
menu_dict = {name: price for name, price in zip(menu_names, prices)}
print("辞書内包表記:", menu_dict)
```
