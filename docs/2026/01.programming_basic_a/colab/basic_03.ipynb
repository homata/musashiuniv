```python
# 変数とデータ型の確認をしてみましょう

item_name = "抹茶ラテ"  # str（文字列）: 記号(" ")で囲む必要があります！
price = 550             # int（整数）: 記号は不要です
tax_rate = 0.1          # float（実数）
is_hot = True           # bool（真偽値）: 最初のTやFは大文字です

print(item_name, "の値段は", price, "円です。")

# 型変換（キャスト）：データの種類を変換することもできます。
# 例えば、数字の 550 を、文字の "550" に変換するには str() を使います。
price_str = str(price)
print("文字列になった値段: " + price_str) # 文字同士なら + 記号で繋げられます！
```

```python
# 税込み価格を計算するオリジナル関数を定義（def）します！

def calc_tax(price):
    # ここからインデントを下げてブロックを作ります
    tax = 0.1  # taxは、この関数の中でしか使えないローカル変数です
    total = price * (1 + tax)

    # 計算結果（完成品）を関数の外に返します！(return)
    return total

# ------------------------------
# ここから関数の外（インデントなし）です

# 関数を呼び出して（材料として 400 を渡す）、結果を変数に入れます
coffee_total = calc_tax(400)
cake_total = calc_tax(500)

print("コーヒーの税込み価格は", coffee_total, "円です。")
print("ケーキの税込み価格は", cake_total, "円です。")

# （補足）関数の参照
# 括弧 () をつけずに関数名だけを変数に入れると、関数そのものに別名をつけることができます。
my_calculator = calc_tax
print("別名で計算:", my_calculator(1000))
```

### 🌟 Colabで試すおすすめコード



```python
# BMI（肥満度）を計算する関数です
def calculate_bmi(weight_kg, height_m):
    # BMI = 体重(kg) ÷ (身長(m) × 身長(m))
    bmi = weight_kg / (height_m * height_m)
    return bmi

# 自分のデータを入れてみましょう（身長はセンチではなくメートルです！）
my_weight = 60.0
my_height = 1.70

my_bmi = calculate_bmi(my_weight, my_height)
print("あなたのBMIは", my_bmi, "です！")
```
