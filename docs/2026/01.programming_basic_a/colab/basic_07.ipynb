```python
# assert文の具体例
def calc_discount(price, discount_rate):
    # 割引率は絶対に 0〜1 の間でなければならない！と断言（assert）します。
    # そうでない場合は、カンマの右側のメッセージを出して停止します。
    assert 0 <= discount_rate <= 1, "エラー：割引率がおかしいです！"

    return price * (1 - discount_rate)

print("正常な計算:", calc_discount(1000, 0.2)) # 20%引き

# 下の行のコメントを外して実行すると、安全装置（assert）が作動して止まります！
# print("異常な計算:", calc_discount(1000, 1.5)) # 150%引き！？
```
