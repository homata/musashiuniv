```python
# 基本的な if / else 文
is_member = True  # 会員かどうか（True または False）
price = 1000

if is_member:
    # ここは is_member が True の時だけ実行されます
    print("会員様ですね。50円割引します！")
    price = price - 50
else:
    # それ以外の時（False の時）はここが実行されます
    print("一般のお客様ですね。通常価格です。")

print("お会計は", price, "円です。")
```

```python
# if / elif / else 文と比較演算子
temperature = 28

if temperature >= 30:
    print("猛暑です！フラペチーノをたくさん仕入れましょう。")
elif temperature >= 25:
    print("夏日です。アイスコーヒーがよく売れそうです。")
else:
    print("涼しいです。ホットコーヒーの準備をしましょう。")

```

```python
# 論理演算子 (and, or)
day = "水曜"
weather = "雨"

# 水曜日「または」雨の日はポイント2倍！
if day == "水曜" or weather == "雨":
    print("本日はポイント2倍デーです！")
```

```python
# None と 空のデータの判定
stock_list = []  # 空のリスト（在庫なし）

# リストが空っぽだと、False 扱いになる性質を利用します！
if not stock_list:
    print("在庫が空っぽです！発注してください。")

# 3項演算子（1行で書くif文）
age = 20
# 年齢が20以上なら"大人"、そうでないなら"未成年"を変数 status に入れる
status = "大人" if age >= 20 else "未成年"
print("このお客様は", status, "です。")
```

### 🌟 Colabで試すおすすめコード


```python
# （準備）売上データファイルを作成します
with open("daily_sales.txt", "w", encoding="utf-8") as f:
    f.write("コーヒー,450\nケーキ,500\nコーヒー,450\n紅茶,400\n")

# コーヒーだけの売上合計を入れる箱
coffee_total = 0

# ファイルを読み込みます
with open("daily_sales.txt", "r", encoding="utf-8") as f:
    lines = f.readlines() # 1行ずつリストにする

# まだ学んでいませんが、次回詳しくやる for文（繰り返し）を使います！
for line in lines:
    clean_line = line.strip()       # 1. 改行などのゴミを消す
    data_list = clean_line.split(",") # 2. カンマで区切ってリストにする

    item_name = data_list # 商品名
    item_price = int(data_list) # 値段（計算したいので int で数値に変換！）

    # 3. もし商品名がコーヒーなら、合計に足す！
    if item_name == "コーヒー":
        coffee_total = coffee_total + item_price

print("本日のコーヒーの売上合計は", coffee_total, "円です！")
```
